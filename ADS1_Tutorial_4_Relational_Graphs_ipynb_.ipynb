```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def sin2(x):
    """
    Calculates sin^2 of x
    """
    f = np.sin(x) ** 2
    return f

def circle():
    """
    Returns x and y values of the unit circle
    """
    phi = np.linspace(0.0, 2.0 * np.pi, 1000)
    x = np.cos(phi)
    y = np.sin(phi)
    return x, y
```

```python
# calculate the sine of ONE value
x = 2.0
y = np.sin(x)
print("y = ", y)

# create an array with 1000 x values from -10 to 10 (inclusive) and printing first 10
# hint: try also using np.arange, instead.
x = np.linspace(-10.0, 10.0, 1000)
print("x =", x[:10])

# calculate the sine for all values at once
y = np.sin(x)
print("y =", y[:10])
```

```python
def plot_sine_cos(x, sinx, cosx):
    """
    Plotting sine and cosine of x array as line graph
    """
    # initialise figure, dpi is the resolution (144 looks good on big screen)
    plt.figure(dpi=144)

    # plot line graph, labels are used to produce the legend
    plt.plot(x, sinx, label="sin(x)")
    plt.plot(x, cosx, label="cos(x)")

    # formatting
    # set the upper and lower limits of x
    plt.xlim(-10.0, 10.0)
    # add axis labels
    plt.xlabel("x")
    plt.ylabel("f(x)")
    # add the legend, which uses the labels
    plt.legend(loc="upper left")
    plt.show()
    return
```

```python
# "call" that function with the created values
plot_sine_cos(x, np.sin(x), np.cos(x))
```

```python
# now with sin^2(x)
def plot_sine_sine2(x, sinx, sin2x):
    """
    Plotting sine and sin^2 of x array as line graph
    """
    # initialise figure, dpi is the resolution (144 looks good on big screen)
    plt.figure(dpi=144)

    # plot line graph, labels are used to produce the legend
    plt.plot(x, sinx, label="sin(x)")
    plt.plot(x, sin2x, label="sin2(x)")

    # formatting
    # set the upper and lower limits of x
    plt.xlim(-10.0, 10.0)
    # add axis labels
    plt.xlabel("x")
    plt.ylabel("f(x)")
    # add the legend, which uses the labels
    plt.legend(loc="upper left")
    plt.show()
    return
```

```python
def plot_sinh_cosh(x, sinh_x, cosh_x):
    plt.figure(dpi=144)
    plt.plot(x, sinh_x, label="sinh(x)", color="blue")
    plt.plot(x, cosh_x, label="cosh(x)", color="red", linestyle="--")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title("Hyperbolic Functions: sinh(x) vs cosh(x)")
    plt.legend(loc="upper center")
    plt.grid(True)
    plt.show()
    return
```

```python
x_hyp = np.linspace(-3.0, 3.0, 500)
plot_sinh_cosh(x_hyp, np.sinh(x_hyp), np.cosh(x_hyp))
```

```python
def plot_trig_polynomial(t, a=1, b=60, c=1, d=120, j=3, k=4):
    x_val = np.cos(a * t) - (np.cos(b * t) ** j)
    y_val = np.sin(c * t) - (np.sin(d * t) ** k)
    plt.figure(dpi=144, figsize=(6, 6))
    plt.plot(x_val, y_val, color="navy", linewidth=0.6)
    plt.xlabel("X(t) = cos(at) - cos(bt)^j")
    plt.ylabel("Y(t) = sin(ct) - sin(dt)^k")
    plt.title("Parametric Trigonometric Polynomial")
    plt.axis("equal")
    plt.grid(True)
    plt.show()
    return
```

```python
t = np.linspace(0, 2 * np.pi, 5000)
plot_trig_polynomial(t)
```

```python
def plot_scatter_with_line(x, y):
    plt.figure(dpi=144)
    plt.scatter(x, y, color="blue", alpha=0.7, label="Data points")
    plt.plot([0, 1], [0, 1], color="black", linewidth=2, linestyle="-", label="Ideal fit (y = x)")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Scatter Plot with Reference Line")
    plt.legend()
    plt.grid(True)
    plt.show()
    return
```

```python
x = np.random.uniform(0, 1, 100)
y = x + np.random.normal(0, 0.2, len(x))
plot_scatter_with_line(x, y)
```
