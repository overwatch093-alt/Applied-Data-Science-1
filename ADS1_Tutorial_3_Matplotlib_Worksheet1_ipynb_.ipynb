```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
rng = np.random.default_rng(42)
print('Libraries ready! NumPy:', np.__version__, '| pandas:', pd.__version__)
```

#Exercise 1 - First Line Plot

```python
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]

plt.plot(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("X vs Y")
plt.show()
```

#Exercise 2 - Style Your Plot

```python
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
plt.plot(x, y, color = "green", linestyle="--", linewidth = 3)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("X vs Y")
plt.show()
```

#Exercise 3 - Multiple Lines & Legend

```python
x = np.arange(0, 6)
y1 = x
y2 = x**2

plt.plot(x, y1, label = "linear", color = "green", linestyle ="--", linewidth = 2)
plt.plot(x, y2, label = "quadratic", color = "red", linestyle =":", linewidth = 5)
plt.title("Linear and Qudratic")
plt.legend()
plt.grid(True)
plt.show()
```

#Exercise 4 - Scatter Plot and Markers

```python
n = 50
x = rng.uniform(0, 10, size=n)
noise = rng.normal(0, 2, size=n)
y = 2*x + 1 + noise
plt.scatter(x, y, marker = "o" ,color = "purple", linestyle = "--", linewidth = 2)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Scatter Plot")
plt.show()
```

#Exercise 5 - Bar Chart

```python
labels = ['Mon','Tue','Wed','Thu','Fri']
values = [12, 18, 9, 14, 20]

plt.bar( labels, values, color = "green", edgecolor = "black")
for i,v in enumerate(values):
  plt.text(i, v + 0.3, str(v), ha="center")
plt.xticks(rotation =20)
plt.show()
```

#Exercise 6 - Histogram

```python
samples = rng.normal(0,1, size =1000)
plt.hist(samples, bins=30, density=True)
plt.axvline(samples.mean(), linestyle='--')
plt.title('Normal(0,1) Histogram')
plt.show()
```

#Exercise 7 - Subplots

```python
x = np.linspace(0, 2*np.pi, 200)
ysin = np.sin(x)
ycos = np.cos(x)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
ax1, ax2 = axes
ax1.plot(x, ysin, color = "green"); ax1.set_title('sin(x)')
ax2.plot(x, ycos, color = "purple"); ax2.set_title('cos(x)')
plt.show()
```

#Exercise 8 - Pie Chart

```python
labels = ['A','B','C','D']
sizes = [30, 25, 20, 25]
explode = [0.05, 0, 0, 0]

plt.pie(sizes, labels = labels,  explode= explode, autopct = "%1.1f%%")
plt.axis('equal')
plt.title('Category Shares')
plt.show()
```

#Exercise 10 - Fill Between (Confidence band)

```python
x = np.linspace(0, 10, 200)
noise = rng.normal(0, 0.2, size=x.size)
y = np.sin(x) + noise
sigma = 0.3

plt.plot(x, y, label='signal')
plt.fill_between(x, y - sigma, y + sigma, alpha=0.3, label='±1σ')
plt.legend()
plt.show()
```
